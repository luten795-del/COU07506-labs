```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = np.arange(-5, 11)

delta = (n == 0).astype(float)    # unit impulse
step = (n >= 0).astype(float)     # unit step
ramp = n * step                  # unit ramp
expo = 0.8**n * step             # real exponential
```

```python
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 21)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for ax, a in zip(axes, a_values):
    x = a**n
    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("x(n)")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
import os
```

```python
fname = "my_voice.wav"

if os.path.exists(fname):
    fs, x = wavfile.read(fname)

    if x.ndim > 1:              # stereo -> mono
        x = x.mean(axis=1)

    x = x / np.max(np.abs(x))    # normalise
else:
    fs = 16000
    t = np.arange(0, 2, 1/fs)
    x = 0.5*np.sin(2*np.pi*440*t) * np.exp(-t)

t = np.arange(len(x)) / fs

print(f"fs = {fs} Hz, duration = {len(x)/fs:.2f} s, samples = {len(x)}")

plt.plot(t, x)
plt.xlabel("time (s)")
plt.ylabel("amplitude")
plt.title("My recording")
plt.show()
```

**What is the sampling rate of your phone's recording? How many numbers are needed to store one minute of it in mono?**

The sampling rate of my phone's recording is 16,000 Hz.

To calculate the number of samples needed to store one minute of mono audio, I use:

N = Sampling Rate × Duration

Since one minute is 60 seconds:

N = 16,000 × 60

N = 960,000 samples.

Therefore, one minute of mono audio requires 960,000 samples.


## Question 2
**Why is a stem plot more appropriate than a continuous line for x(n)?**

A stem plot is more appropriate for a discrete-time signal because the signal is defined only at specific integer sample indices. It clearly displays individual sample values and their positions. A continuous line may suggest that values exist between the discrete sample points.

## Question 3
**For a = 1.1 the exponential grows. Which property of a system would such a signal violate if it were an impulse response?**

It would violate the BIBO (Bounded-Input, Bounded-Output) stability property. For a discrete-time system to be BIBO stable, its impulse response must be absolutely summable. Since the exponential signal grows continuously, its absolute sum is infinite. Therefore, the system would be unstable.
